## Introductie

Dit notebook is het eerste notebook van de casus. Start hier dus met het algemene doel van de casus. Benoem daarbij met welke dataset je gaat werken, wat je gaat onderzoeken en wat jouw onderzoeksvraag is. Beschrijf ook kort wat je verwacht te vinden. Na dat algemene doel kun je het doel van dit notebook beschrijven: beschrijf wat je gaat doen qua data understanding en data preparation. 


## 1. Data inlezen & beschrijven

Lees de (ruwe) data in. Bekijk hoe groot je dataset is en welke kolommen er zijn. Maak een data dictionary als die er nog niet is. Beschrijf de kolommen en hun datatype. Benoem wat de features zijn en wat de target variabele is. Benoem bij de features ook welke geschikt zijn om te gebruiken en aan welke een ethisch risico kleeft. Bekijk de eerste paar rijen van de dataset en beschrijf wat je ziet. Beschrijf opvallende zaken. 

## 2. Datakwaliteit

Controleer hier of de datatypes kloppen, of er vreemde waarden in je dataset zitten, of er missende waarden zijn of misschien dubbele rijen.

Gebruik hierbij het stappenplan met de vier stappen zoals beschreven in de introductie.

### Gebruik extra kopjes
Je mag extra kopjes toevoegen om bepaalde acties / onderwerpen te scheiden. Probeer voor een zo overzichtelijk mogelijk notebook te zorgen.

## 3. Data opschonen

Wanneer er bij stap 3 zaken qua datakwaliteit naar voren zijn gekomen die van invloed zijn op de vervolgstappen, is het belangrijk om de datakwaliteit meteen te verbeteren door een aantal essentiële opschoningsacties uit te voeren. Denk er aan om ook hier weer het stappenplan te gebruiken.

**Merk op**: je mag de opschoningsacties ook meteen doorvoeren bij de analyse bij punt 2. In dat geval kun je dit kopje verwijderen, pas dan wel de nummering aan.

## 4. Univariate verkenning met visualisaties

Onderzoek de inhoud van individuele variabelen. Denk aan het visualiseren van de verdeling van de inhoud (hiermee kun je ook outliers detecteren) of alle unieke waarden. Probeer er wel voor te waken om gigantische hoeveelheden data te printen of niet-zeggende visualisaties te plotten. Zorg altijd voor een goed verhaal.

## 5. Bivariate & multivariate verkenning

Onderzoek, waar relevant, relaties tussen variabelen door middel van visualisaties of correlaties. Onderzoek ook het effect van outliers op bijvoorbeeld de correlatie.

## 6. Data preparation

Bij de data understanding acties van punt 5 en 6 zijn ongetwijfeld zaken naar voren gekomen waarmee de data verbeterd kan worden. Voer die verbeteracties hier uit.


## 7. Resultaten & conclusie

Breng je bevindingen samen en benoem de belangrijkste inzichten. 

Wanneer je de data opgeschoond heb, exporteer dan de opgeschoonde dataset zodat deze in het volgende notebook gebruikt kan worden.

## 8. Reflectie & vervolg

Reflecteer kort op wat je hebt gedaan. Wat verraste je? Wat vond je lastig? Wat zou je met meer tijd of data nog onderzoeken?
