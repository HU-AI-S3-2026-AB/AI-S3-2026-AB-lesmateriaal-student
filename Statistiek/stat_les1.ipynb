## Kansen

De kans op een gebeurtenis is een getal tussen de 0 en de 1 dat aangeeft hoe waarschijnlijk het is dat een gebeurtenis A zich voordoet.
Een kans van 0 geeft aan dat een gebeurtenis zeker niet plaatsvindt, bij een kans van 1 vindt die zeker wel plaats. De notatie voor de kans op gebeurtenis A is: $P(A)$.

**Voorbeelden:**

Stel $A$ is de gebeurtenis: het regent morgen. Dan betekent $P(A) = 0.25$ dat het morgen waarschijnlijk niet gaat regenen, maar dat we er niet helemaal zeker van zijn.

Stel $B$ is de gebeurtenis: we gooien een zuivere dobbelsteen en de uitkomst is 6. Dan schrijven we $P(B) = \frac{1}{6}$.

Stel $C$ is de gebeurtenis dat het morgen regent of niet regent, dan $P(C) = 1$.

Stel $D$ is de gebeurtenis dat morgen al het water ineens rood is, dan $P(D) = 0$.

## Stochasten

De $X$ in het voorbeeld van de dobbelsteen is een **stochast** (of *random variable*). Dit is een variabele waarvan de waarde afhangt van het toeval.
Als $X$ de uitkomst van een worp met een dobbelsteen aangeeft, dan geldt:

$$P(X=1) = P(X=2) = P(X=3) = P(X=4) = P(X=5) = P(X=6) = \frac{1}{6}$$

Merk op dat als je al deze kansen optelt je op 1 uitkomt. Dit geldt altijd bij een stochast.

## Rekenen met kansen

**Optelregel voor elkaar uitsluitende gebeurtenissen:**

Als slechts één van de gebeurtenissen $A_1, A_2, \ldots, A_n$ tegelijk kan voorkomen, dan geldt:
$$P(A_1 \vee A_2 \vee \ldots \vee A_n) = P(A_1) + P(A_2) + \cdots + P(A_n)$$

Voorbeeld: de kans op een even getal met een dobbelsteen:
$$P(X=2 \vee X=4 \vee X=6) = \frac{1}{6} + \frac{1}{6} + \frac{1}{6} = 0.5$$

**Optelregel voor overlappende gebeurtenissen:**

Als twee gebeurtenissen $A_1$ en $A_2$ een overlap hebben:
$$P(A_1 \vee A_2) = P(A_1) + P(A_2) - P(A_1 \wedge A_2)$$

**Vermenigvuldigingsregel voor onafhankelijke gebeurtenissen:**

Als $A$ en $B$ onafhankelijk zijn (ze beïnvloeden elkaar niet):
$$P(A \wedge B) = P(A) \cdot P(B)$$

Voorbeeld: Jason en Roos gooien elk met een dobbelsteen. De kans dat *beiden* 4 of meer gooien:
$$P(A \wedge B) = 0.5 \cdot 0.5 = 0.25$$

## Kernel Density Estimation (KDE) en de kansdichtheidsgrafiek

De KDE-lijn in het histogram geeft een vloeiende schatting van de onderliggende **verdeling** van de data.
Als je veel meer data zou verzamelen en de bins heel klein zou maken, dan zou het histogram steeds meer op de KDE-lijn gaan lijken.

Er zijn een aantal veelvoorkomende vormen voor deze verdeling. We behandelen er in deze les drie:
- **Uniforme verdeling**
- **Normale verdeling**
- **Lognormale verdeling**

## Visualiseren van een uniforme verdeling

Bij een uniforme verdeling heeft iedere waarde binnen een interval dezelfde kans. Het gooien van een dobbelsteen is een voorbeeld.
We simuleren dit met NumPy:

```python
# Simuleer 100.000 dobbelsteenworpen
eyes = np.random.randint(1, 7, size=100000)
sns.histplot(eyes, stat='probability', bins=6)
plt.title('Uniforme verdeling: dobbelsteenworpen')
plt.xlabel('Uitkomst')
plt.ylabel('Kans')
plt.show()
```

## De normale verdeling

Een andere veelvoorkomende continue kansverdeling is de **normaalverdeling**. Deze kom je in de natuur veel tegen (lengtes, gewichten) en ook in veel features van datasets. De kansdichtheidsgrafiek is **klokvormig**: de meeste waarden liggen rondom het gemiddelde en de kans neemt af voor waarden verder van het gemiddelde.

De normaalverdeling wordt volledig beschreven door 2 parameters:
- Het **gemiddelde** $\mu$: bepaalt waar de top van de klokvorm ligt
- De **standaarddeviatie** $\sigma$: bepaalt hoe breed of smal de klokvorm is

De kansdichtheidsfunctie is:
$$f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x - \mu)^2}{2\sigma^2}}$$

```python
# Genereer data uit drie verschillende normaalverdelingen
mu_1, sigma_1 = 170, 5
mu_2, sigma_2 = 180, 10
mu_3, sigma_3 = 180, 25

x = np.linspace(100, 300, 1000)

def normal_distribution(mu, sigma, x):
    return 1/(sigma * np.sqrt(2 * np.pi)) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

plt.figure(figsize=(10, 6))
plt.plot(x, normal_distribution(mu_1, sigma_1, x), label=f'N(170, 5)')
plt.plot(x, normal_distribution(mu_2, sigma_2, x), label=f'N(180, 10)')
plt.plot(x, normal_distribution(mu_3, sigma_3, x), label=f'N(180, 25)')
plt.title('Normale verdelingen met verschillende mu en sigma')
plt.xlabel('x')
plt.ylabel('Kansdichtheid')
plt.legend()
plt.show()
```

## Vuistregels voor de normale verdeling

De normaalverdeling heeft een aantal handige eigenschappen (de **68-95-99.7 regel**):
1. Ongeveer **68%** van de waarden ligt binnen $\mu \pm \sigma$.
2. Ongeveer **95%** van de waarden ligt binnen $\mu \pm 2\sigma$.
3. Ongeveer **99.7%** van de waarden ligt binnen $\mu \pm 3\sigma$.

## Kansen berekenen met de normale verdeling

Als een variabele normaal verdeeld is, kunnen we de kans op een interval **exact** berekenen.
We gebruiken hiervoor de **cumulatieve verdelingsfunctie** (`cdf`).

`stats.norm.cdf(t, loc=mu, scale=sigma)` geeft de kans dat een variabele $\leq t$ is — dus de oppervlakte links van $t$ onder de kansdichtheidsgrafiek.

![normale verdeling](./../afbeeldingen/Data/normaleverdeling.png "Kansdichtheidsgrafiek normale verdeling")

Voorbeeld: de snavellengte van een pinguïnsoort is normaal verdeeld met $\mu = 44$ mm en $\sigma = 5$ mm.
Hoe groot is de kans dat een willekeurige pinguïn een snavel heeft die langer is dan 50 mm?

```python
import scipy.stats as stats

kans_kleiner_dan_50 = stats.norm.cdf(50, loc=44, scale=5)
kans_groter_dan_50 = 1 - kans_kleiner_dan_50
print(f"Kans dat een pinguïn een snavel heeft langer dan 50mm: {kans_groter_dan_50:.2f}")
```

## De z-score en datapreparatie

Als je weet dat je data normaal verdeeld is, kun je de **z-score** gebruiken. Dit is een maat die aangeeft hoeveel standaarddeviaties een bepaalde waarde van het gemiddelde afligt:

$$z = \frac{x - \mu}{\sigma}$$

Z-scores zijn nuttig voor:
1. **Outlier detectie**: waardes met $|z| > 3$ zijn mogelijke outliers.
2. **Normalisatie**: alle waarden op dezelfde schaal brengen (gemiddelde = 0, std = 1). Dit is belangrijk voor sommige machine learning modellen zoals kNN (zie de ML-lessen).

In les 2 van de statistiek-leerlijn zal de z-score (of een variatie daarop) ook terugkomen als basis voor de t-toets.

```python
# Voorbeeld: z-scores voor de lengte van Europese vrouwen
length = np.array([1.59, 1.60, 1.67, 1.71, 1.71, 1.76, 1.68, 1.83, 1.65, 1.95, 1.74,
       1.59, 1.57, 1.73, 1.66, 1.75, 1.73, 1.67, 1.60, 1.53, 1.64, 1.77,
       1.70, 1.56, 1.70, 1.72, 1.59, 1.70, 1.69, 1.57])

mean_length = np.mean(length)
std_length = np.std(length)
z_scores = (length - mean_length) / std_length
print(f"Gemiddelde: {mean_length:.2f} m")
print(f"Standaarddeviatie: {std_length:.2f} m")
print(f"\nZ-scores:\n{z_scores.round(2)}")
print(f"\nMogelijke outliers (|z| > 3): {length[np.abs(z_scores) > 3]}")
```

## Lognormale verdeling

Een andere veelvoorkomende verdeling in de praktijk is de **lognormale verdeling**. Dit is een scheve verdeling met een lange staart naar rechts. Je ziet deze vaak bij prijzen, inkomens en andere variabelen die altijd positief zijn en waarbij grote uitschieters voorkomen.

Door het **logaritme** van de waarden te nemen, krijg je een normale verdeling — dit kan handig zijn voor visualisatie en outlier detectie.

```python
# Genereer lognormale data
data_lognormal = np.random.lognormal(mean=0, sigma=0.5, size=10_000)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Oorspronkelijke lognormale verdeling
axes[0].hist(data_lognormal, bins=40, edgecolor='k', alpha=0.7)
axes[0].set_title('Lognormale verdeling')
axes[0].set_xlabel('Waarde')
axes[0].set_ylabel('Frequentie')

# Na log-transformatie: normale verdeling
data_log = np.log(data_lognormal)
axes[1].hist(data_log, bins=40, edgecolor='k', alpha=0.7)
axes[1].set_title('Na log-transformatie: normale verdeling')
axes[1].set_xlabel('log(Waarde)')
axes[1].set_ylabel('Frequentie')

plt.tight_layout()
plt.show()
```
